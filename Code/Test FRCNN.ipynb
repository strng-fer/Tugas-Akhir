## 3. Custom Dataset Class untuk COCO Format

```python
def load_coco_json(json_path):
    """Load COCO format JSON annotation file"""
    with open(json_path, 'r') as f:
        return json.load(f)

def get_classes_from_coco(data_dir):
    """Extract all class names from COCO JSON annotations"""
    json_path = os.path.join(data_dir, '_annotations.coco.json')

    if not os.path.exists(json_path):
        print(f"⚠️ Warning: {json_path} not found!")
        return [], {}, {}, {}

    coco_data = load_coco_json(json_path)

    # Extract categories
    categories = coco_data.get('categories', [])
    class_names = [cat['name'] for cat in categories]

    # Create mappings (COCO id -> our id, +1 for background)
    coco_id_to_class_name = {cat['id']: cat['name'] for cat in categories}
    class_to_idx = {cat['name']: idx + 1 for idx, cat in enumerate(categories)}
    idx_to_class = {idx + 1: cat['name'] for idx, cat in enumerate(categories)}
    idx_to_class[0] = '__background__'

    # Create COCO ID to our ID mapping
    coco_id_to_idx = {cat['id']: class_to_idx[cat['name']] for cat in categories}

    return class_names, class_to_idx, idx_to_class, coco_id_to_idx

# Get class names from training data

print("\n" + "="*80)
print("LOADING COCO ANNOTATIONS")
print("="*80)

CLASS_NAMES, class_to_idx, idx_to_class, coco_id_to_idx = get_classes_from_coco(CONFIG['train_dir'])

if len(CLASS_NAMES) > 0:
    print(f"\nTotal Classes: {len(CLASS_NAMES)}")
    print(f"Total classes (with background): {len(class_to_idx) + 1}")

    # Update CONFIG with actual number of classes
    actual_num_classes = len(CLASS_NAMES) + 1  # +1 for background
    if actual_num_classes != CONFIG['num_classes']:
        print(f"\n⚠️ Updating num_classes from {CONFIG['num_classes']} to {actual_num_classes}")
        CONFIG['num_classes'] = actual_num_classes

    print("\nClass Names:")
    for idx, name in enumerate(CLASS_NAMES, 1):
        print(f"  {idx}. {name}")

    # Validate label mapping
    print(f"\n✓ Label Validation:")
    print(f"  Label range: 1 to {max(coco_id_to_idx.values())}")
    print(f"  Expected range: 1 to {CONFIG['num_classes']-1}")
    print(f"  COCO category IDs: {sorted(coco_id_to_idx.keys())}")

    if max(coco_id_to_idx.values()) >= CONFIG['num_classes']:
        print(f"\n❌ ERROR: Label indices exceed num_classes!")
        print(f"  Max label: {max(coco_id_to_idx.values())}")
        print(f"  num_classes: {CONFIG['num_classes']}")
        raise ValueError("Label mapping error - will cause CUDA assert errors during training")
else:
    print("\n⚠️ ERROR: No classes found!")
    print("Please check:")
    print(f"  1. File exists: {os.path.join(CONFIG['train_dir'], '_annotations.coco.json')}")
    print(f"  2. File contains 'categories' field")
    print(f"  3. Dataset was downloaded and extracted correctly")
    raise ValueError("No classes found in dataset")
```

```python
import os
import json
import shutil
from sklearn.model_selection import train_test_split, KFold
from collections import defaultdict
import numpy as np

print("="*80)
print("5-FOLD CROSS VALIDATION WITH ONLINE AUGMENTATION")
print("="*80)
print("Configuration:")
print("  Outer Split (TrainVal:Test): 85:15")
print("  Inner Split (Train:Val per fold): 80:20")
print("  Augmentation: Online (during training)")
print("  Model Types: Online augmentation pada setiap batch training")
print("="*80)

# Define paths based on CONFIG
original_train_dir = CONFIG['train_dir']
original_valid_dir = CONFIG['valid_dir']
original_test_dir = CONFIG['test_dir']
output_base_dir = CONFIG['base_dir']

# New CV directories
cv_base_dir = os.path.join(output_base_dir, 'cv_folds')
test_cv_dir = os.path.join(cv_base_dir, 'test')

# Ensure output directory is clean
if os.path.exists(cv_base_dir):
    print(f"Removing existing CV directory: {cv_base_dir}")
    shutil.rmtree(cv_base_dir)
os.makedirs(cv_base_dir, exist_ok=True)
os.makedirs(test_cv_dir, exist_ok=True)

print(f"\nOriginal data directories:")
print(f"  Train: {original_train_dir}")
print(f"  Valid: {original_valid_dir}")
print(f"  Test: {original_test_dir}")

# ================================================================================
# STEP 1: Combine all training and validation data
# ================================================================================
print("\n" + "="*80)
print("STEP 1: LOAD & COMBINE ALL DATA (Train + Valid + Test from Roboflow)")
print("="*80)

def load_coco_data(json_path):
    """Load COCO annotation file"""
    with open(json_path, 'r') as f:
        return json.load(f)

# Load data from all three directories
all_images = []
all_annotations = []
img_id_offset = 0
img_id_mapping = {}  # old_id -> new_id

categories = None

for data_dir, dir_name in [
    (original_train_dir, 'train'),
    (original_valid_dir, 'valid'),
    (original_test_dir, 'test')
]:
    json_path = os.path.join(data_dir, '_annotations.coco.json')

    if not os.path.exists(json_path):
        print(f"⚠️ Warning: {json_path} not found, skipping...")
        continue

    print(f"\nLoading {dir_name} data...")
    coco_data = load_coco_data(json_path)

    if categories is None:
        categories = coco_data['categories']

    print(f"  Images: {len(coco_data['images'])}")
    print(f"  Annotations: {len(coco_data['annotations'])}")

    # Remap image IDs to be unique across all datasets
    for img in coco_data['images']:
        old_id = img['id']
        new_id = img_id_offset + old_id
        img_id_mapping[(dir_name, old_id)] = new_id
        img['id'] = new_id
        all_images.append(img)

    # Remap annotation image_ids
    for ann in coco_data['annotations']:
        old_img_id = ann['image_id']
        new_img_id = img_id_offset + old_img_id
        ann['image_id'] = new_img_id
        all_annotations.append(ann)

    img_id_offset += max([img['id'] for img in coco_data['images']]) + 1000

print(f"\nTotal combined data:")
print(f"  Total images: {len(all_images)}")
print(f"  Total annotations: {len(all_annotations)}")
print(f"  Categories: {len(categories)}")

# Create mapping of image_id to image info and annotations
img_id_to_info = {img['id']: img for img in all_images}
img_id_to_anns = defaultdict(list)
for ann in all_annotations:
    img_id_to_anns[ann['image_id']].append(ann)

# ================================================================================
# STEP 2: OUTER SPLIT - 85% TrainVal, 15% Test
# ================================================================================
print("\n" + "="*80)
print("STEP 2: OUTER SPLIT (TrainVal:Test = 85:15)")
print("="*80)

all_image_ids = np.array(list(img_id_to_info.keys()))
np.random.seed(42)

# Split: 85% for CV, 15% for final test set
trainval_ids, test_ids = train_test_split(
    all_image_ids,
    test_size=0.15,
    random_state=42
)

print(f"TrainVal set (85%): {len(trainval_ids)} images")
print(f"Test set (15%): {len(test_ids)} images")

# Create test split for later evaluation
def create_coco_split(image_ids_subset, output_dir, img_id_to_info, img_id_to_anns,
                      categories, original_data_dirs, img_id_mapping_reverse):
    """Create COCO split with copied images"""
    new_images = []
    new_annotations = []

    for img_id in image_ids_subset:
        if img_id not in img_id_to_info:
            continue

        img_info = img_id_to_info[img_id].copy()
        new_images.append(img_info)

        # Copy image file (find which original directory it came from)
        found = False
        for data_dir in original_data_dirs:
            src_path = os.path.join(data_dir, img_info['file_name'])
            if os.path.exists(src_path):
                dst_path = os.path.join(output_dir, os.path.basename(img_info['file_name']))
                shutil.copy(src_path, dst_path)
                img_info['file_name'] = os.path.basename(img_info['file_name'])
                found = True
                break

        if not found:
            print(f"  Warning: Image file not found for {img_info['file_name']}")

        # Copy annotations
        new_annotations.extend([ann.copy() for ann in img_id_to_anns[img_id]])

    new_coco_data = {
        'images': new_images,
        'annotations': new_annotations,
        'categories': categories,
        'info': {'description': 'Plankton Dataset - COCO Format'},
        'licenses': []
    }

    os.makedirs(output_dir, exist_ok=True)
    output_json = os.path.join(output_dir, '_annotations.coco.json')
    with open(output_json, 'w') as f:
        json.dump(new_coco_data, f, indent=2)

    return len(new_images), len(new_annotations)

# Save test split
original_data_dirs = [original_train_dir, original_valid_dir, original_test_dir]
test_imgs, test_anns = create_coco_split(test_ids, test_cv_dir, img_id_to_info,
                                         img_id_to_anns, categories,
                                         original_data_dirs, img_id_mapping)
print(f"\nTest split created: {test_imgs} images, {test_anns} annotations")

# ================================================================================
# STEP 3: 5-FOLD CROSS VALIDATION on TrainVal (80:20 per fold)
# ================================================================================
print("\n" + "="*80)
print("STEP 3: 5-FOLD CROSS VALIDATION (Train:Val = 80:20 per fold)")
print("="*80)

NUM_FOLDS = 5
kf = KFold(n_splits=NUM_FOLDS, shuffle=True, random_state=42)

fold_info = {}

for fold_idx, (train_indices, val_indices) in enumerate(kf.split(trainval_ids)):
    fold_name = f"fold_{fold_idx + 1}"
    fold_dir = os.path.join(cv_base_dir, fold_name)

    train_fold_dir = os.path.join(fold_dir, 'train')
    val_fold_dir = os.path.join(fold_dir, 'val')

    os.makedirs(train_fold_dir, exist_ok=True)
    os.makedirs(val_fold_dir, exist_ok=True)

    # Get image IDs for this fold
    train_fold_ids = trainval_ids[train_indices]
    val_fold_ids = trainval_ids[val_indices]

    # Create splits
    train_count, train_anns = create_coco_split(train_fold_ids, train_fold_dir,
                                                img_id_to_info, img_id_to_anns,
                                                categories, original_data_dirs,
                                                img_id_mapping)

    val_count, val_anns = create_coco_split(val_fold_ids, val_fold_dir,
                                            img_id_to_info, img_id_to_anns,
                                            categories, original_data_dirs,
                                            img_id_mapping)

    fold_info[fold_idx] = {
        'fold_name': fold_name,
        'fold_dir': fold_dir,
        'train_dir': train_fold_dir,
        'val_dir': val_fold_dir,
        'train_count': train_count,
        'val_count': val_count,
        'train_anns': train_anns,
        'val_anns': val_anns
    }

    print(f"\n{fold_name}:")
    print(f"  Train: {train_count} images ({100*train_count/(train_count+val_count):.1f}%)")
    print(f"  Val:   {val_count} images ({100*val_count/(train_count+val_count):.1f}%)")

# Store CV configuration
print(f"\n{'='*80}")
print(f"5-Fold CV Structure Created at: {cv_base_dir}")
print(f"{'='*80}\n")

# Save fold information as JSON for reference
fold_info_json = os.path.join(cv_base_dir, 'fold_info.json')
fold_summary = {fold_idx: {k: v for k, v in info.items() if k != 'fold_dir' and k != 'train_dir' and k != 'val_dir'}
                for fold_idx, info in fold_info.items()}
with open(fold_info_json, 'w') as f:
    json.dump(fold_summary, f, indent=2)

print(f"Fold information saved to: {fold_info_json}")

# Store fold info for later use
CV_FOLD_INFO = fold_info
CV_TEST_DIR = test_cv_dir
print(f"\nCV_FOLD_INFO dictionary created for later use in training loop")
```

```python
from sklearn.model_selection import StratifiedKFold

# Prepare labels for stratified splitting of trainval_ids.
# For each image in trainval_ids, we get all its associated unique category_ids.
# These are then sorted and joined into a string to create a unique signature
# for the combination of classes present in that image.
# This string will serve as the 'y' input for StratifiedKFold.
# Images without any annotations are assigned a special 'no_objects' label.
print("\n" + "="*80)
print("PREPARING LABELS FOR STRATIFIED 5-FOLD CV")
print("="*80)

trainval_labels_for_stratification = []
for img_id in trainval_ids:
    annotations = img_id_to_anns.get(img_id, [])
    categories_in_image = sorted(list(set(ann['category_id'] for ann in annotations)))

    if not categories_in_image:
        # Assign a special label for images without objects (if any)
        trainval_labels_for_stratification.append('no_objects')
    else:
        # Create a unique string signature for the combination of classes present in the image
        trainval_labels_for_stratification.append('_'.join(map(str, categories_in_image)))

# Convert the list of labels to a numpy array for StratifiedKFold
trainval_labels_np = np.array(trainval_labels_for_stratification)

print("\n✅ Labels for stratified K-Fold prepared.")
print(f"  Number of trainval images: {len(trainval_ids)}")
print(f"  Number of labels for stratification: {len(trainval_labels_np)}")
print(f"  Unique label combinations: {len(np.unique(trainval_labels_np))}")

# Check for label combinations that might cause StratifiedKFold to fail (i.e., fewer samples than folds)
unique_labels, counts = np.unique(trainval_labels_np, return_counts=True)
single_sample_labels = unique_labels[counts < NUM_FOLDS] # NUM_FOLDS is defined in cell ba2d9a1f

if len(single_sample_labels) > 0:
    print("\n⚠️ WARNING: The following label combinations have fewer samples than NUM_FOLDS:")
    for label in single_sample_labels:
        print(f"  - '{label}' (count: {counts[unique_labels == label][0]})")
    print("  StratifiedKFold might raise an error if these labels cannot be distributed across all folds.")
    print("  Consider inspecting your dataset for very rare label combinations or using a non-stratified split if errors occur.")
else:
    print("  All label combinations have sufficient samples for stratification.")

print("\n" + "="*80)
```

```python
from torchvision import transforms

# Define the augmentation pipeline for training data
# This is a basic placeholder; you can customize it further.
AUGMENTATION_TRAIN = transforms.Compose([
    transforms.ToTensor(), # Convert PIL image to PyTorch Tensor
    # Add your desired augmentations here, for example:
    # transforms.RandomHorizontalFlip(p=0.5),
    # transforms.RandomVerticalFlip(p=0.5),
    # transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1),
    # transforms.RandomRotation(degrees=15),
    # transforms.RandomResizedCrop(CONFIG['input_size'], scale=(0.8, 1.0), ratio=(0.75, 1.333)),
    # Note: For object detection, geometric transforms need to be applied carefully to bounding boxes
    # (often handled within a custom dataset or augmentation library like Albumentations).
    # The PlanktonCOCODataset class itself might handle some of these with bbox adjustments.
])

# Custom Dataset class for COCO format
class PlanktonCOCODataset(Dataset):
    def __init__(self, root_dir, coco_id_to_idx, input_size, is_train=True, augmentation=None):
        self.root_dir = root_dir
        self.coco_id_to_idx = coco_id_to_idx
        self.input_size = input_size
        self.is_train = is_train
        self.augmentation = augmentation

        self.json_path = os.path.join(root_dir, '_annotations.coco.json')
        if not os.path.exists(self.json_path):
            raise FileNotFoundError(f"COCO annotation file not found: {self.json_path}")

        with open(self.json_path, 'r') as f:
            self.coco = json.load(f)

        self.img_ids = [img['id'] for img in self.coco['images']]
        self.img_id_to_info = {img['id']: img for img in self.coco['images']}
        self.img_id_to_annotations = defaultdict(list)
        for ann in self.coco['annotations']:
            self.img_id_to_annotations[ann['image_id']].append(ann)

        # Define default transformations if no augmentation is provided
        self.default_transform = transforms.Compose([
            transforms.ToTensor(), # Convert PIL image to PyTorch Tensor
            transforms.Resize((self.input_size, self.input_size)) # Resize to model input size
        ])

    def __len__(self):
        return len(self.img_ids)

    def __getitem__(self, idx):
        img_id = self.img_ids[idx]
        img_info = self.img_id_to_info[img_id]
        annotations = self.img_id_to_annotations[img_id]

        img_path = os.path.join(self.root_dir, img_info['file_name'])
        image = Image.open(img_path).convert("RGB")

        boxes = []
        labels = []
        for ann in annotations:
            x, y, w, h = ann['bbox']
            boxes.append([x, y, x + w, y + h]) # Convert to [x_min, y_min, x_max, y_max]
            labels.append(self.coco_id_to_idx[ann['category_id']])

        boxes = torch.as_tensor(boxes, dtype=torch.float32)
        labels = torch.as_tensor(labels, dtype=torch.int64)

        # Create target dictionary
        target = {}
        target["boxes"] = boxes
        target["labels"] = labels
        target["image_id"] = torch.tensor([img_id])
        target["area"] = (boxes[:, 3] - boxes[:, 1]) * (boxes[:, 2] - boxes[:, 0])
        target["iscrowd"] = torch.zeros((len(boxes),), dtype=torch.int64)

        # Apply augmentation for training, or default transform for validation/testing
        if self.is_train and self.augmentation:
            # Note: For object detection, augmentation needs to handle bounding boxes.
            # This placeholder assumes an augmentation pipeline that can handle both image and target.
            # A more robust solution might use Albumentations or custom augmentation logic.
            # For now, we'll apply image-only transforms and hope for the best, or rely on dataset-specific handling.
            # If the augmentation modifies image size, bbox coordinates will need adjustment.
            # For simplicity, we'll apply augmentation after resizing and then adjust bboxes.

            # Resize image first to the expected input_size for the model
            image = transforms.Resize((self.input_size, self.input_size))(image)

            # Apply augmentation (if it only affects image pixels, it's fine)
            image = self.augmentation(image)

            # NOTE: If your augmentation modifies the geometry of the image (e.g., flip, rotate, crop),
            # you MUST adjust the bounding box coordinates accordingly. This simple setup does not
            # automatically adjust bounding boxes for geometric transforms in `self.augmentation`.
            # A full implementation would pass both `image` and `target` to the augmentation step.

        else:
            # Apply default transformations (resize and to_tensor)
            original_width, original_height = image.size
            image = self.default_transform(image)

            # Adjust bounding boxes based on resize operation for validation/test
            scale_x = self.input_size / original_width
            scale_y = self.input_size / original_height
            target["boxes"][:, 0] *= scale_x
            target["boxes"][:, 1] *= scale_y
            target["boxes"][:, 2] *= scale_x
            target["boxes"][:, 3] *= scale_y

        return image, target

def collate_fn(batch):
    """Custom collate function for DataLoader"""
    return tuple(zip(*batch))

print("="*80)
print("5-FOLD CROSS VALIDATION DATA LOADING")
print("="*80)

def get_cv_dataloaders(fold_idx, cv_fold_info, augmentation_train, num_workers=0, batch_size=1):
    """
    Create train/val dataloaders for a specific fold with online augmentation

    Args:
        fold_idx: Fold index (0-4)
        cv_fold_info: Dictionary containing fold information
        augmentation_train: Augmentation pipeline for training
        num_workers: Number of workers for DataLoader
        batch_size: Batch size

    Returns:
        Tuple of (train_loader, val_loader, train_dataset, val_dataset)
    """

    fold_info = cv_fold_info[fold_idx]
    train_dir = fold_info['train_dir']
    val_dir = fold_info['val_dir']

    print(f"\nCreating dataloaders for {fold_info['fold_name']}...")

    # Create train dataset with augmentation
    print(f"  Creating training dataset with online augmentation...")
    train_dataset = PlanktonCOCODataset(
        root_dir=train_dir,
        coco_id_to_idx=coco_id_to_idx,
        input_size=CONFIG['input_size'],
        is_train=True,
        augmentation=augmentation_train  # Online augmentation applied here!
    )

    # Create validation dataset (no augmentation)
    print(f"  Creating validation dataset (no augmentation)...")
    val_dataset = PlanktonCOCODataset(
        root_dir=val_dir,
        coco_id_to_idx=coco_id_to_idx,
        input_size=CONFIG['input_size'],
        is_train=False,
        augmentation=None
    )

    # Create dataloaders
    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        num_workers=num_workers,
        collate_fn=collate_fn,
        pin_memory=True if torch.cuda.is_available() else False
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers,
        collate_fn=collate_fn,
        pin_memory=True if torch.cuda.is_available() else False
    )

    print(f"  ✓ Train loader: {len(train_dataset)} samples")
    print(f"  ✓ Val loader: {len(val_dataset)} samples")

    return train_loader, val_loader, train_dataset, val_dataset

def get_test_dataloader(test_dir, num_workers=0, batch_size=1):
    """
    Create test dataloader

    Args:
        test_dir: Path to test directory
        num_workers: Number of workers
        batch_size: Batch size

    Returns:
        test_loader, test_dataset
    """
    print(f"\nCreating test dataset...")

    test_dataset = PlanktonCOCODataset(
        root_dir=test_dir,
        coco_id_to_idx=coco_id_to_idx,
        input_size=CONFIG['input_size'],
        is_train=False,
        augmentation=None
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers,
        collate_fn=collate_fn,
        pin_memory=True if torch.cuda.is_available() else False
    )

    print(f"  ✓ Test loader: {len(test_dataset)} samples")

    return test_loader, test_dataset

# ================================================================================
# EXAMPLE: Create dataloaders for Fold 1 (you can iterate through all folds)
# ================================================================================
print("\n" + "="*80)
print("CREATING DATALOCADERS FOR FOLD 1 (example)")
print("="*80)

FOLD_IDX = 0  # First fold (Fold 1)

train_loader, val_loader, train_dataset, val_dataset = get_cv_dataloaders(
    fold_idx=FOLD_IDX,
    cv_fold_info=CV_FOLD_INFO,
    augmentation_train=AUGMENTATION_TRAIN,
    num_workers=CONFIG['num_workers'],
    batch_size=CONFIG['batch_size']
)

# Also create test loader
test_loader, test_dataset = get_test_dataloader(
    test_dir=CV_TEST_DIR,
    num_workers=CONFIG['num_workers'],
    batch_size=CONFIG['batch_size']
)

print(f"\n{'='*80}")
```

```python
import glob
import random
from torchvision.models.detection import FasterRCNN, fasterrcnn_mobilenet_v3_large_fpn
from torchvision.models.detection.backbone_utils import resnet_fpn_backbone

# -----------------------------------------------------------------------------
# Load trained checkpoints from Google Drive
# -----------------------------------------------------------------------------
MODEL_WEIGHT_CANDIDATES = {
    'ResNet50': ['/content/drive/MyDrive/Best Model/resnet50.pth'],
    'ResNet101': [
        '/content/drive/MyDrive/Best Model/resnet5101.pth',
        '/content/drive/MyDrive/Best Model/resnet101.pth',
        '/content/drive/MyDrive/Best Model/resnet50_101.pth',
    ],
    'MobileNetV3-Large': ['/content/drive/MyDrive/Best Model/mobilenetv3.pth'],
}

TEST_VIS_DIR = os.path.join('outputs_faster_rcnn', 'test_visualizations')
os.makedirs(TEST_VIS_DIR, exist_ok=True)

def build_faster_rcnn(backbone_name, num_classes):
    """Build a Faster R-CNN model that matches the trained backbone."""
    backbone_name = backbone_name.lower()
    if backbone_name == 'resnet50':
        model = torchvision.models.detection.fasterrcnn_resnet50_fpn(
            weights=None,
            weights_backbone=None,
            num_classes=num_classes,
        )
    elif backbone_name == 'resnet101':
        backbone = resnet_fpn_backbone(
            'resnet101',
            weights=None,
            trainable_layers=3,
        )
        model = FasterRCNN(backbone, num_classes=num_classes)
    elif backbone_name == 'mobilenet_v3_large':
        model = fasterrcnn_mobilenet_v3_large_fpn(
            weights=None,
            weights_backbone=None,
            num_classes=num_classes,
        )
    else:
        raise ValueError(f'Unsupported backbone: {backbone_name}')
    return model

def resolve_weight_path(candidates, model_name):
    """Return the first existing checkpoint path, or search the folder for a model-specific match."""
    for candidate in candidates:
        if os.path.exists(candidate):
            return candidate

    search_dir = os.path.dirname(candidates[0])
    if os.path.isdir(search_dir):
        all_candidates = sorted(glob.glob(os.path.join(search_dir, '*.pth')))
        if len(all_candidates) == 1:
            return all_candidates[0]

        model_name_lower = model_name.lower()
        if 'mobilenet' in model_name_lower:
            expected_tokens = ['mobilenet', 'v3', 'large']
        elif '101' in model_name_lower:
            expected_tokens = ['resnet', '101']
        else:
            expected_tokens = ['resnet', '50']

        for path in all_candidates:
            normalized = os.path.basename(path).lower().replace('-', '').replace('_', '').replace('.pth', '')
            if all(token in normalized for token in expected_tokens):
                return path

        # Final fallback: if there are multiple files and one clearly contains the model family name.
        for path in all_candidates:
            lower_name = os.path.basename(path).lower()
            if 'mobilenet' in model_name_lower and 'mobile' in lower_name:
                return path
            if '101' in model_name_lower and '101' in lower_name:
                return path
            if '50' in model_name_lower and '50' in lower_name:
                return path

    raise FileNotFoundError(f'No matching weight file found. Candidates: {candidates}')

def extract_state_dict(checkpoint):
    """Handle checkpoints saved as a raw state_dict or wrapped dictionary."""
    if hasattr(checkpoint, 'state_dict') and not isinstance(checkpoint, dict):
        return None, checkpoint
    if isinstance(checkpoint, dict):
        for key in ['model_state_dict', 'state_dict', 'model', 'net']:
            value = checkpoint.get(key)
            if isinstance(value, dict):
                checkpoint = value
                break
        if any(key.startswith('module.') for key in checkpoint.keys()):
            checkpoint = {key.replace('module.', '', 1): value for key, value in checkpoint.items()}
        return checkpoint, None
    return None, checkpoint

def load_trained_model(weight_path, backbone_name, num_classes, device):
    if not os.path.exists(weight_path):
        raise FileNotFoundError(f'Weight file not found: {weight_path}')

    checkpoint = torch.load(weight_path, map_location='cpu', weights_only=False)
    state_dict, full_model = extract_state_dict(checkpoint)

    if full_model is not None:
        model = full_model
    else:
        model = build_faster_rcnn(backbone_name, num_classes)
        model.load_state_dict(state_dict, strict=True)

    model.to(device)
    model.eval()
    return model

print('=' * 90)
print('LOADING TRAINED MODELS')
print('=' * 90)

loaded_models = {}
for model_name, candidates in MODEL_WEIGHT_CANDIDATES.items():
    weight_path = resolve_weight_path(candidates, model_name)
    if 'ResNet101' in model_name:
        backbone_name = 'resnet101'
    elif 'ResNet50' in model_name:
        backbone_name = 'resnet50'
    else:
        backbone_name = 'mobilenet_v3_large'

    print(f'Loading {model_name} from {weight_path}')
    loaded_models[model_name] = load_trained_model(
        weight_path=weight_path,
        backbone_name=backbone_name,
        num_classes=CONFIG['num_classes'],
        device=CONFIG['device'],
    )
    print(f'  -> {model_name} ready on {CONFIG["device"]}')

print('\n✓ All models loaded successfully')

# -----------------------------------------------------------------------------
# Prepare test dataset and helpers
# -----------------------------------------------------------------------------
test_dataset = test_dataset

def tensor_to_image(image_tensor):
    image = image_tensor.detach().cpu().permute(1, 2, 0).numpy()
    image = np.clip(image, 0, 1)
    return image

def draw_boxes(ax, image, boxes, labels, scores=None, title=None, color='lime', threshold=0.5):
    ax.imshow(image)
    ax.axis('off')
    if title is not None:
        ax.set_title(title, fontsize=11)

    for idx, box in enumerate(boxes):
        score = None if scores is None else float(scores[idx])
        if score is not None and score < threshold:
            continue

        x1, y1, x2, y2 = box
        label_idx = int(labels[idx]) if labels is not None else 0
        class_name = idx_to_class.get(label_idx, str(label_idx))
        label_text = class_name if score is None else f'{class_name} {score:.2f}'

        rect = plt.Rectangle(
            (x1, y1),
            x2 - x1,
            y2 - y1,
            fill=False,
            edgecolor=color,
            linewidth=2,
        )
        ax.add_patch(rect)
        ax.text(
            x1,
            max(0, y1 - 4),
            label_text,
            color='white',
            fontsize=9,
            bbox=dict(facecolor=color, alpha=0.8, pad=2, edgecolor='none'),
        )

def run_inference(model, image_tensor, score_threshold=0.5):
    with torch.no_grad():
        prediction = model([image_tensor.to(CONFIG['device'])])[0]
    boxes = prediction['boxes'].detach().cpu().numpy()
    labels = prediction['labels'].detach().cpu().numpy()
    scores = prediction['scores'].detach().cpu().numpy()
    keep = scores >= score_threshold
    return boxes[keep], labels[keep], scores[keep]

# Use a few deterministic test images so the comparison is repeatable
num_samples = min(40, len(test_dataset))
if num_samples == 0:
    raise ValueError('Test dataset is empty. Please check CONFIG["test_dir"] and the COCO annotations.')

random.seed(42)
sample_indices = list(range(len(test_dataset)))[:num_samples]

figure, axes = plt.subplots(
    nrows=num_samples,
    ncols=len(loaded_models) + 1,
    figsize=(5 * (len(loaded_models) + 1), 5 * num_samples),
    squeeze=False,
 )

print('\n' + '=' * 90)
print('RUNNING TEST INFERENCE AND VISUALIZATION')
print('=' * 90)

for row, sample_idx in enumerate(sample_indices):
    image_tensor, target = test_dataset[sample_idx]
    image_np = tensor_to_image(image_tensor)

    gt_boxes = target['boxes'].detach().cpu().numpy()
    gt_labels = target['labels'].detach().cpu().numpy()

    draw_boxes(
        axes[row, 0],
        image_np,
        gt_boxes,
        gt_labels,
        scores=None,
        title=f'Test sample {sample_idx} - Ground Truth',
        color='deepskyblue',
        threshold=0.0,
    )

    for col, (model_name, model) in enumerate(loaded_models.items(), start=1):
        pred_boxes, pred_labels, pred_scores = run_inference(
            model,
            image_tensor,
            score_threshold=0.5,
        )
        draw_boxes(
            axes[row, col],
            image_np,
            pred_boxes,
            pred_labels,
            scores=pred_scores,
            title=f'{model_name} - {len(pred_boxes)} deteksi',
            color='lime',
            threshold=0.5,
        )

plt.tight_layout()
output_path = os.path.join(TEST_VIS_DIR, 'test_comparison_grid.png')
plt.savefig(output_path, dpi=200, bbox_inches='tight')
plt.show()

print(f'\n✓ Visualisasi disimpan ke: {output_path}')
print('✓ Inference selesai untuk tiga model pada data test terpilih')
```
